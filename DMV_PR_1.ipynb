```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Load CSV, Excel and JSON files
df_csv = pd.read_csv("supermarket_sales - Sheet1 (1).csv")

df_excel = pd.read_excel("Superstore (1).xlsx")

df_json = pd.read_json("StoreSales (1).json", encoding_errors="ignore")
```

```python

# Remove duplicate records
df_csv.drop_duplicates(inplace=True)
df_excel.drop_duplicates(inplace=True)
df_json.drop_duplicates(inplace=True)

# Fill missing values
df_csv['Total'] = df_csv['Total'].fillna(df_csv['Total'].mean())

df_csv['Gender'] = df_csv['Gender'].fillna(df_csv['Gender'].mode()[0])

# Remove incomplete JSON sales records
df_json = df_json.dropna(subset=['Category', 'Sales', 'Quantity'])
```

```python

df_csv_data = pd.DataFrame({
    'Category': df_csv['Product line'],
    'Sales': df_csv['Total'],
    'Quantity': df_csv['Quantity']
})

df_excel_data = pd.DataFrame({
    'Category': df_excel['Category'],
    'Sales': df_excel['Sales'],
    'Quantity': df_excel['Quantity']
})

df_json_data = pd.DataFrame({
    'Category': df_json['Category'],
    'Sales': df_json['Sales'],
    'Quantity': df_json['Quantity']
})

# Combine all datasets
df = pd.concat([df_csv_data, df_excel_data, df_json_data],
               ignore_index=True)

df.head()
```

```python

# Convert date into datetime format
df_csv['Date'] = pd.to_datetime(df_csv['Date'])

# Extract year and month
df_csv['Year'] = df_csv['Date'].dt.year
df_csv['Month'] = df_csv['Date'].dt.month

# Calculate amount
df_csv['Calculated Amount'] = df_csv['Unit price'] * df_csv['Quantity']

df_csv.head()
```

```python

# Group sales by region
region_sales = df_excel.groupby('Region')['Sales'].sum()

region_sales
```

```python

# Bar plot
region_sales.plot(kind='bar', figsize=(10, 6))

plt.title('Sales Distribution by Region')
plt.xlabel('Region')
plt.ylabel('Total Sales')
plt.show()
```

```python

# Pie chart
region_sales.plot(kind='pie', autopct='%1.1f%%')

plt.title('Sales Distribution by Region')
plt.ylabel('')
plt.show()
```
